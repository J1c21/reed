## Interactive charts

```python
import plotly.express as px
df = px.data.gapminder().query("year == 2007")
fig = px.scatter(df, x="gdpPercap", y="lifeExp", size="pop", color="continent", log_x=True, height=380)
fig
```

```python
import altair as alt
source = pd.DataFrame({"x": np.arange(60), "y": np.sin(np.arange(60) / 6)})
alt.Chart(source).mark_line(point=True).encode(x="x", y="y").properties(width=520, height=220)
```

## Live updates

```python
import time
handle = display(Markdown("starting…"), display_id=True)
for i in range(1, 6):
    time.sleep(0.15)
    handle.update(Markdown(f"**progress** {'█' * i}{'░' * (5 - i)} {i * 20}%"))
```

```python
from IPython.display import clear_output
for i in range(4):
    clear_output(wait=True)
    print(f"frame {i}")
    time.sleep(0.15)
```

```python
from tqdm.auto import tqdm
for _ in tqdm(range(40), desc="widgets"):
    time.sleep(0.02)
```

```python
import ipywidgets as w
@w.interact(n=(1, 20), color=["indigo", "teal", "crimson"], shout=False)
def demo(n=5, color="indigo", shout=False):
    text = f"{n} squared is {n * n}"
    print(text.upper() if shout else text, "in", color)
```
